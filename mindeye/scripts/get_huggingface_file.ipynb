```python
from huggingface_hub import hf_hub_download
import os, shutil

# ------------------ SETTINGS ------------------
repo_id = "macandro96/mindeye_offline_ckpts"
# ckpt_subpath = "sub-005_ses-01_task-C_bs24_MST_rishab_MSTsplit_1_avgrepeats_finalmask_epochs_15/last.pth"
ckpt_subpath = "sub-005_ses-01_task-C_bs24_MST_rishab_MSTsplit_3_avgrepeats_finalmask_epochs_45/last.pth"

# final desired file name:
final_name = ckpt_subpath.replace('/last','')

gcs_dir  = "/teamspace/gcs_folders/share/real_time_mindeye_data/3t_data/data/model"
local_tmp = os.path.expanduser("~/.hf_tmp")
# ------------------------------------------------


os.makedirs(local_tmp, exist_ok=True)

print("downloading from HF to local tmp ...")
orig_path = hf_hub_download(
    repo_id=repo_id,
    filename=ckpt_subpath,
    local_dir=local_tmp,
    local_dir_use_symlinks=False,
    force_download=True,
    resume_download=False,
)
print("orig_path:", orig_path)

final_path = os.path.join(gcs_dir, final_name)

print("copying to GCS final location ...")
shutil.copy2(orig_path, final_path)

print("\nDONE.")
print("saved as:", final_path)
print("size:", os.path.getsize(final_path))
```
